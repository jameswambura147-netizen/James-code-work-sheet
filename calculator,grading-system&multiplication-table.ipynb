```python
##simple calculations
print("**********SIMPLE CALCULATOR**********")
print("Enter any two numbers:")
num1=float(input("Enter first number: "))
num2=float(input("Enter second number: "))

add=num1+num2
sub=abs(num1-num2)
mult=num1*num2
div=num1/num2
mod=num1%num2

print("==============RESULTS============")
print(num1 , "+" , num2 ,"=", add)
print(num1 , "-" , num2 ,"=", sub)
print(num1 , "x" , num2 ,"=", mult)
print(num1 , "/" , num2 ,"=", div)
print("Remainder btn ", num1 ,"/" ,num2,"=", mod)
print("===================================")
```

```python
##grading system
num=int(input("Enter number of students you have: "))
sub=int(input("Enter number of subjects: "))
results=[]
for i in range(num):
    name=input(f"Enter full name of student{i+1}: ")
    total=0
    for j in range(sub):
        marks=float(input(f"Enter marks of subject{j+1}: "))
        total+=marks
    results.append(f"{name} = {total}")
    print()
print("*************RESULTS**************")
for result in results:
    print(results)
print("***********************************")
```

```python
##sample 2 calculator
def calcu(num1,num2,sign):
        if sign == '+':
            return num1+num2
        elif sign == '-':
            if num1>num2:
                return num1-num2
            else:
                return f'- {num2-num1}'
        elif sign == '/':
            if num2==0:
                return 'division error'
            else:
                return num1/num2
        elif sign =='*':
            return num1*num2
        else:
            return 'Invalid operator'

a=float(input('Enter first number: '))
b=float(input('Enter second number: '))
sn=input('Enter an operator or ''x'' to exit: ')
result=calcu(a,b,sn)
print('****************************')
print(f'Answer of {a} {sn} {b} = {result}')
```

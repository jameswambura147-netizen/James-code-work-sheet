```python
def grade(marks):
    if marks>80:
        return 'A'
    elif marks>70:
        return 'B'
    elif marks>60:
        return 'C'
    elif marks>45:
        return 'D'
    else:
        return 'F'

num=int(input('number of students: '))
for i in range(num):
    name=input(f'Enter name of student {i+1} : ')
    score=float(input('Enter marks:'))
    result=grade(score) #calling a function by passing to it an argument
    
    print(f'{name} has got {score} which is grade {result}')
    print('-----------------------------------------------')
    print()
```

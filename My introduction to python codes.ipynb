```python
email="jam.es@12.com"
email=email.strip()
if email=="":
    print("The emsil can not be empty")
elif not '.' in email or not '@' in email:
    print("Email must contain '.' and '@'")
elif email.count('@') != 1:
    print("Email must contain atleast one '@'")
elif not email.endswith(('.com','.org','.net')):
    print("Email must end with .com,.org or.net")
elif len(email)>254:
    print("An email must not be longer than 254 characters")
#isalnum()  checks if the string contains only letters and digits
elif not email[0].isalnum() and email[-1].isalnum():
    print("must start with a letter or digit")
else:
    print("Email is Valid;")
```

```python
password=input("Enter your password: ")
email=input("Enter your email: ")
if password=="":
    print("Password must be filled")
elif len(password)<8:
    print("Password must contain atleast 8 characters")
#any() means is atleast one of them true
#c just a character in the password
elif not any(c.isupper() for c in password):
    print("Must include atleast one uppercase")
elif not any(c.islower() for c in password):
    print("Must include atleast one lowercase")
elif password == email :
    print("Your password should not be the same as email")
elif " " in password:
    print("Your password Must not contain any spaces")
elif not (password[0].isalnum() and password[-1].isalnum()):
    print("Password must start with a letter or digit")
else:
    print("Valid credentials:")
```

```python
from numpy import *
from matplotlib.pyplot import *

xstart=0
xstop=6
increment=0.2
x=arange(xstart,xstop,increment)

y=sin(x)
z=cos(x)
subplot(2,1,1)
plot(x,y,'g')
title('sin')
xlabel('x')
ylabel('sin(x)')
grid()
show()

subplot(2,1,2)
plot(x,z,'r')
title('cos')
xlabel('x')
ylabel('cos(x)')

show()
```

```python
#Functions
def least_difference(a,b,c):
    diff1=abs(a-b)
    diff2=abs(b-c) 
    diff3=abs(a-c)
    return min(diff1,diff2,diff3)

print(
    least_difference(1,10,100),
    least_difference(1,10,10),
    least_difference(5,6,7)
)
```

```python
def mult(x):
    return 5*x
def call(fn,arg):
    return fn(arg)
print(
    mult(1),
    call(mult,2),
    sep='\n'
)
```

```python
def modulo(x):
    return x%5
print(
    'Which one is the biggest: ',
    max(144,23,345),
    'Biggest module of 5: ',
    max(34,25,48, key=modulo),
    sep='\n'
)
```

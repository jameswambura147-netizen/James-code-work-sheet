```python
##simple calculator
def calcu(num1,num2,sign):
        if sign == '+':
            return num1+num2
        elif sign == '-':
            if num1>num2:
                return num1-num2
            else:
                return f'- {num2-num1}'
        elif sign == '/':
            if num2==0:
                return 'division error'
            else:
                return num1/num2
        elif sign =='*':
            return num1*num2
        else:
            return 'Invalid operator'

a=float(input('Enter first number: '))
b=float(input('Enter second number: '))
sn=input('Enter an operator : ')
result=calcu(a,b,sn)
print('****************************')
print(f'Answer of {a} {sn} {b} = {result}')
```

```python
#swapping
Men="stronger"
Women="More stronger"

print("Before Swapping")
print("=======================")
print("Men= ",Men)
print("Women= ",Women)
print("=======================")

#swapping
Sex="" #just an intermediate
Sex=Men
Men=Women
Women=Sex

print()
print("After Swapping")
print("=======================")
print("Men= ",Men)
print("Women= ",Women)
print("=======================")
```
